```python
import re
```

```python
DLM = "|"
with open("../projekt_data/Hungarian-NRC-EmoLex.txt", "r", encoding="utf-8") as file:
    first_line = re.sub(r"	", DLM, file.readline())
    print("Header:", first_line)
    
    data_lines = []
    for line in file:
        line = re.sub(r"	", DLM, line)
        data_lines.append(line)
```

```python
print(first_line)

header_item_list = first_line.strip().split(DLM)
print(header_item_list)

data_lines_list = [line.strip().split(DLM) for line in data_lines]
print(data_lines_list)
```

```python
data = {}
items_dropped = 0

for data_item in data_lines_list:
    if len(data_item) != len(header_item_list):
        print("Data item length mismatch:", data_item)
        items_dropped += 1
        continue
    
    eng_word = data_item[0]
    translation = data_item[len(data_item) - 1]
    word_value_dict = {}

    for i in range(1, len(header_item_list)-1):
        word_value_dict[header_item_list[i]] = data_item[i]

    try:
        total_value = sum(int(value) for value in word_value_dict.values() if value.isdigit())
        if total_value == 0:
            print(f"Total value is zero for data item: {eng_word}")
            items_dropped += 1
            continue
    except ValueError:
        print(f"Value conversion error in data item: {eng_word}")

    data[eng_word] = {
        "translation": translation,
        "values": word_value_dict
    }

    print("Data item added:", word_value_dict)
    print("English word:", eng_word)
    print("Translation:", translation)

print("Total items processed:", len(data))
print("Items dropped:", items_dropped)
```

```python
print(data["abacus"]["values"]["trust"]) 
```

```python
def get_emotions(eng_word):
    if eng_word in data:
        values = data[eng_word]["values"]
        available_emotions = [key for key, value in values.items() if value == '1']
        return available_emotions
    else:
        print(f"Word '{eng_word}' not found in data.")
        return None
    
get_emotions("abacus")
```
